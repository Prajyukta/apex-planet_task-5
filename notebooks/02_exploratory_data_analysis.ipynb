# 02 Exploratory Data Analysis
Profile revenue, profitability, customer activity, and regional/category patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('../data/processed/master_cleaned_data.csv', parse_dates=['Order_Date'])
daily = data.groupby('Order_Date', as_index=False).agg(Revenue=('Total_Revenue','sum'), Profit=('Net_Profit','sum'))
daily.plot(x='Order_Date', y=['Revenue','Profit'], marker='o', title='Daily Revenue and Profit')
plt.tight_layout()
```

```python
kpis = data.agg(Revenue=('Total_Revenue','sum'), Profit=('Net_Profit','sum'), Active_Customers=('Customer_ID','nunique'))
kpis
```
